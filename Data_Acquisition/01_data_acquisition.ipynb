```python
from pathlib import Path

def find_project_root(start: Path) -> Path:
    start = start.resolve()

    required_any = [
        ["dataset", "README.md"],
        ["dataset", "notebooks"],
        ["dataset", "data_acquisition"],
        ["dataset", "preprocessing"],
    ]

    for p in [start, *start.parents]:
        for markers in required_any:
            if all((p / m).exists() for m in markers):
                return p

    raise FileNotFoundError(f"Could not find project root from {start}")
```

```python
NOTEBOOK_DIR = Path.cwd().resolve()
PROJECT_ROOT = find_project_root(NOTEBOOK_DIR)

DATASET_DIR = PROJECT_ROOT / "dataset"
# TRAINING_DIR = PROJECT_ROOT / "training"
DATA_DIR = PROJECT_ROOT / "dataset" / "dataM5stickC"
APP_SCRIPT = PROJECT_ROOT / "data_acquisition"
# DIST_DIR = PROJECT_ROOT / "dist"
```

The next cell triggers **a GUI --> lets you scan for available bluetooth devices --> connect --> visualize and/or store** activity data.. 

(**Please** be assured to flash the firmware before executing this step otherwise this step doesn't make any sense..

**Note:** The location of the data collected is controlled by the python script (imu_collector.py), change if necessary..

```python
import subprocess
import sys
from pathlib import Path

def get_app_candidates(project_root: Path):
    dist_dir = project_root / "dist"

    if sys.platform.startswith("win"):
        return [
            dist_dir / "HAR Data Acquisition.exe",
            dist_dir / "HAR_Data_Acquisition.exe",
        ]
    elif sys.platform == "darwin":
        return [
            dist_dir / "HAR Data Acquisition.app" / "Contents" / "MacOS" / "HAR Data Acquisition",
            dist_dir / "HAR_Data_Acquisition.app" / "Contents" / "MacOS" / "HAR_Data_Acquisition",
        ]
    elif sys.platform.startswith("linux"):
        return [
            dist_dir / "HAR Data Acquisition",
            dist_dir / "HAR_Data_Acquisition",
        ]
    else:
        return []

app_candidates = get_app_candidates(PROJECT_ROOT)
existing_app = next((p for p in app_candidates if p.exists()), None)

script_candidates = [
    NOTEBOOK_DIR / "imu_collector.py",
    NOTEBOOK_DIR / "imucollector.py",
    NOTEBOOK_DIR / "studio" / "imu_collector.py",
    NOTEBOOK_DIR / "data_acquisition" / "imucollector.py",
]
existing_script = next((p for p in script_candidates if p.exists()), None)

print(f"Notebook dir : {NOTEBOOK_DIR}")
print(f"Project root : {PROJECT_ROOT}")
print(f"Platform     : {sys.platform}")

if existing_app:
    print(f"Found standalone app: {existing_app}")
    subprocess.Popen([str(existing_app)])
elif existing_script:
    print(f"Found Python app script: {existing_script}")
    subprocess.Popen([sys.executable, str(existing_script)])
else:
    print("ERROR: No runnable app was found.")
    print("\nChecked standalone app candidates:")
    for p in app_candidates:
        print(f" - {p}")
    print("\nChecked Python script candidates:")
    for p in script_candidates:
        print(f" - {p}")

print("\nCollect your data in the app, glance repo for further steps !!")
```
